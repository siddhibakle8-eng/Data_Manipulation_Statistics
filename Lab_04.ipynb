```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/content/Retail_Dataset_Modified.csv")

print("Dataset Loaded Successfully")
```

```python
print("Data Types:")
print(df.dtypes)
```

```python
df.info()
```

```python
df['Quantity_Available'] = pd.to_numeric(
    df['Quantity_Available'],
    errors='coerce'
)
```

```python
print(df['Quantity_Available'].dtypes)
```

```python
df['Units_Sold'] = pd.to_numeric(
    df['Units_Sold'],
    errors='coerce'
)

print(df['Units_Sold'].dtypes)
```

```python
df['Revenue'] = pd.to_numeric(
    df['Revenue'],
    errors='coerce'
)

print(df['Revenue'].dtypes)
```

```python
df['Shipping_Cost'] = pd.to_numeric(
    df['Shipping_Cost'],
    errors='coerce'
)

print(df['Shipping_Cost'].dtypes)
```

```python
print(df[['Quantity_Available',
          'Units_Sold',
          'Revenue',
          'Shipping_Cost']].dtypes)
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
missing_cols = df.isnull().sum()
print(
    missing_cols[missing_cols > 0]
)
```

```python
df['City'].fillna(
    df['City'].mode()[0],
    inplace=True
)
```

```python
df['Product_Name'].fillna(
    df['Product_Name'].mode()[0],
    inplace=True
)
```

```python
df['Quantity_Available'].fillna(
    df['Quantity_Available'].mean(),
    inplace=True
    )
```

```python
df['Units_Sold'].fillna(
    df['Units_Sold'].mean(),
    inplace=True
    )
```

```python
df['Revenue'].fillna(
    df['Revenue'].mean(),
    inplace=True
    )
```

```python
df['Supplier'].fillna(
    df['Supplier'].mode()[0],
    inplace=True
    )
```

```python
df['Customer_Satisfaction'].fillna(
    df['Customer_Satisfaction'].mean(),
    inplace=True
    )
```

```python
df['Delivery_Time'].fillna(
    df['Delivery_Time'].mean(),
    inplace=True
    )
```

```python
df['Shipping_Cost'].fillna(
    df['Shipping_Cost'].mean(),
    inplace=True
    )
```
