```python
revenue_skew = df["Revenue"].skew()
print(revenue_skew)
```

```python
if revenue_skew > 0:
    print("Positive Skew")
elif revenue_skew < 0:
    print("Negative Skew")
else:
    print("Symmetric Distribution")
```

```python
skewness=df["Revenue"].skew()

if skewness>0:
  print("The positive skewed distribution:",skewness)
elif skewness<0:
  print("The negative skewed distribution")
else:
  print("The symmetric distribution")
```

```python
units_skew = df["Units_Sold"].skew()
print(units_skew)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(df["Revenue"], bins=20)
plt.title("Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.hist(df["Units_Sold"], bins=20)
plt.title("Units Sold Distribution")
plt.xlabel("Units Sold")
plt.ylabel("Frequency")
plt.show()
```

```python
summary = pd.DataFrame({
    "Metric":["Revenue","Units Sold"],
    "Skewness":[revenue_skew,units_skew]
})
summary
```
